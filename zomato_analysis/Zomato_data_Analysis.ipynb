Dropping The Columns

```python
cols = ['Locality Verbose','Rating color','Rating text','Switch to order menu','Latitude','Longitude']
```

```python
df.drop(cols,axis = 1, inplace = True)

df.columns
```

```python
df.head()
```

```python
df['Cuisines'].head()
```

```python
# remove exrta whitespaces 

df['Cuisines'] = df['Cuisines'].str.strip().str.replace(', ',',')
```

```python
df['Cuisines'].head()
```

# we'd split into a list and then explode our dataframe to have only one Cuisines per row for ezch restaurant.

```python
df['Cuisines'] = df['Cuisines'].str.split(', ')

df = df.explode('Cuisines').reset_index(drop = True)
```

```python
df.head()
```

```python
df['Address'].head()
```

```python
# remove newline from Address column

df['Address'].str.replace(r'\\n','')
```

```python
df.shape
```

```python
df.info()
```

```python
# drop missing values

df.dropna(subset = ['Cuisines'],inplace = True)
```
